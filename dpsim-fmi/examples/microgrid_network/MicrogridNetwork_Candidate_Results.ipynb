# OpenModelica microgrid FMU candidate

This notebook documents the imported candidate without inventing a simulation result. It plots the state and interface inventory from `sources/config.json`, and marks the platform/interface limitations that must be resolved before native DPsim EMT coupling.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt

root = Path.cwd()
if not (root / "sources").exists():
    root = root / "dpsim-fmi" / "examples" / "microgrid_network"
config = json.loads((root / "sources" / "config.json").read_text())
config
```

```python
labels = ["continuous states", "FMI current inputs", "candidate voltages"]
values = [
    config["continuous_states"],
    len(config["current_inputs"]),
    len(config["candidate_voltage_outputs"]),
]
fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(labels, values, color=["#4472c4", "#70ad47", "#ed7d31"])
ax.set_ylabel("count")
ax.set_title("Imported microgrid candidate interface inventory")
ax.bar_label(bars)
fig.tight_layout()
plt.show()
```

## Interpretation

The candidate is substantially more complex than the local test FMUs, but it is not a drop-in `NonlinearSSN` component. Its upstream binaries are Linux/Windows only, and the useful voltage signals are local variables. The next implementation step would be a deliberate current-driven adapter with an explicit local-variable output mapping; this notebook does not claim that adapter already exists.
