# Small grid: post-PF DPsim and Simulink comparison

This notebook compares fresh closed-loop DPsim ME, DPsim CS, native Simulink, and Simulink CS runs after an independently calculated balanced AC power-flow initialization. Every case uses a 5 s run, the same timestep within each comparison, and a 50 MW PCC load step at 2.5 s.

Simulink initialization was solved independently from the one-bus AC equations for the source, PI feeder, shunt capacitance, and the inverter's 60 Hz periodic current. No DPsim initial values or recorded traces were used to initialize Simulink. The 25 µs ME run is not used as the reference here: native Simulink is the error reference for each matched timestep.


````python
from pathlib import Path

roots = [Path.cwd(), *Path.cwd().parents]
repo = next(
    root
    for root in roots
    if (
        root
        / "build-fmi/small-inverter-grid/results/post_powerflow_fix/simulink_comparison/figures"
    ).is_dir()
)
figures = (
    repo
    / "build-fmi/small-inverter-grid/results/post_powerflow_fix/simulink_comparison/figures"
)
print(f"Using compact saved plots in {figures}")
````

## Reproduce

The script independently computes the AC startup point, builds a Simulink-CS-specific copy of the FMU with matching start values, runs 50 ms smoke cases first, then the 5 s native and CS sweeps, and regenerates compact metrics and figures:

```bash
python dpsim-fmi/examples/small_inverter_grid/run_post_powerflow_simulink_comparison.py
```

The full-resolution traces, timing records, metrics, and figures are under `build-fmi/small-inverter-grid/results/post_powerflow_fix/simulink_comparison/`. The customized FMU is staged separately under `build-fmi/`; the checked-in FMU sources are not altered by this comparison runner.
