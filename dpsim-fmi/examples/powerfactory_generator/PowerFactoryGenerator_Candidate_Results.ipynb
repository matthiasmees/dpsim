# PowerFactory/Modelica generator candidate

This notebook visualizes the imported model inventory and documents the intended use for operating-point initialization. It does not fabricate an EMT result for the Windows-only upstream binary.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt

root = Path.cwd()
if not (root / "sources").exists():
    root = root / "dpsim-fmi" / "examples" / "powerfactory_generator"
config = json.loads((root / "sources" / "config.json").read_text())
config
```

```python
labels = ["continuous states", "inputs", "outputs"]
values = [config["continuous_states"], len(config["inputs"]), len(config["outputs"])]
fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(labels, values, color=["#4472c4", "#70ad47", "#ed7d31"])
ax.set_ylabel("count")
ax.set_title("GENROU/SCRX candidate interface inventory")
ax.bar_label(bars)
fig.tight_layout()
plt.show()
```

## Coupling interpretation

`vi` and `vr` are the real and imaginary voltage-interface inputs, while `p` and `q` are the exported power outputs. A future power-flow initializer can use this interface directly at the operating-point level, but an abc EMT adapter and a rollback strategy are still required for the network solver.
