# IEEE9 DPsim inverter comparison: native, FMI ME, and FMI CS

This notebook compares three fresh, closed-loop DPsim runs of the same IEEE9 EMT network. The averaged inverter is at bus 3, the network starts from the case powerflow, and a 25 MW load is switched onto bus 8 at 5 s. Each method runs for 10 s at the same communication/network timestep. Native DPsim is the error baseline for ME and CS at each matching timestep. The standard IEEE9 PiLine, transformer, and RX-load subcomponents are enabled.

The native averaged inverter uses the same P/Q target, current limit, passive current sign, and 5 ms current response as the dual-interface FMU. Its initial current state is set to the FMU's start behavior (the P/Q reference evaluated at the solved powerflow voltage). Errors are computed from full-resolution run logs before those large files are discarded; compact plotting traces are retained. All ME cases in this regenerated sweep use residual-based Newton convergence with a 20-iteration ceiling.

**Regenerate runs** from the repository root with `python3 -u dpsim-fmi/examples/ieee9_inverter/run_ieee9_dpsim_native_me_cs_sweep.py`. The sweep includes 10, 25, 50, 100, 500, 1000, and 5000 µs; 5000 µs is shown as a coarse stress case. The notebook only reads saved metrics and traces.

> **Stale saved CS results:** The plots and metrics below were generated with the earlier single-pass Co-Simulation path and are not the corrected IEEE9 CS results. The IEEE9 runner now maps `cs` to rollback/Newton; `cs-explicit` preserves the old one-pass mode. Use [the corrected 5 s rollback/Newton validation](IEEE9_CoSimulation_RollbackNewton_Validation.ipynb) for current CS results, or rerun this notebook after regenerating its data.

```python
from pathlib import Path
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

root = Path.cwd().resolve()
while root != root.parent and not (root / "dpsim-fmi").is_dir():
    root = root.parent
if not (root / "dpsim-fmi").is_dir():
    raise FileNotFoundError("Open this notebook from a DPsim checkout")
results_dir = (
    root / "build-fmi" / "ieee9-inverter" / "dpsim-native-me-cs-10s-full-components"
)
metrics_path = results_dir / "metrics.json"
metrics = json.loads(metrics_path.read_text())
print(f"Results: {metrics_path}")
print(metrics["configuration"])
```

## Full-run errors and timing

```python
rows = []
for step, case in metrics["cases"].items():
    for method in ("native", "me", "cs"):
        r = case["runtimes"][method]
        comp = case["comparisons"].get(method)
        s = comp["full_run"] if comp else {}
        rows.append(
            {
                "timestep [µs]": int(step),
                "method": method.upper(),
                "PCC voltage RMS error [%]": s.get("voltage_rms_error_percent"),
                "inverter current RMS error [%]": s.get("current_rms_error_percent"),
                "solver wall [s]": r["solver_wall_s"],
                "process wall [s]": r["process_wall_s"],
            }
        )
summary = pd.DataFrame(rows).sort_values(["timestep [µs]", "method"])
display(
    summary.style.format(
        {c: "{:.5g}" for c in summary.select_dtypes("number").columns}, na_rep="—"
    )
)
fig, ax = plt.subplots(1, 3, figsize=(18, 4))
for method, group in summary.groupby("method"):
    if method != "NATIVE":
        ax[0].plot(
            group["timestep [µs]"],
            group["PCC voltage RMS error [%]"],
            "o-",
            label=method,
        )
        ax[1].plot(
            group["timestep [µs]"],
            group["inverter current RMS error [%]"],
            "o-",
            label=method,
        )
    ax[2].plot(group["timestep [µs]"], group["solver wall [s]"], "o-", label=method)
for a, title, ylabel in zip(
    ax,
    [
        "PCC voltage error vs native DPsim",
        "Inverter current error vs native DPsim",
        "Solver wall time",
    ],
    [
        "Full-run three-phase RMS error [%]",
        "Full-run three-phase RMS error [%]",
        "Run time [s]",
    ],
):
    a.set_xscale("log")
    a.set_xlabel("Fixed step [µs]")
    a.set_ylabel(ylabel)
    a.set_title(title)
    a.grid(True, alpha=0.3)
    a.legend()
plt.tight_layout()
```

## Initialization check

```python
init_rows = []
for step, case in metrics["cases"].items():
    for method, comp in case["comparisons"].items():
        x = comp["initial"]
        init_rows.append(
            {
                "step [µs]": int(step),
                "method": method.upper(),
                "native Vabc at t=0 [V]": x["native_v_abc_V"],
                "candidate Vabc at t=0 [V]": x["candidate_v_abc_V"],
                "native Iabc at t=0 [A]": x["native_i_abc_A"],
                "candidate Iabc at t=0 [A]": x["candidate_i_abc_A"],
            }
        )
display(pd.DataFrame(init_rows))
```

## Per-timestep error traces and load-step response

For every timestep, the first figure shows the trailing 100 ms three-phase RMS error. The second shows cumulative RMS error from time zero through each point, estimated from the retained compact traces; the full-run summary above uses full-resolution data. The lower waveform figure zooms to ±25 ms around the 5 s event and overlays phase-a PCC voltage and inverter current for all three methods.

```python
for step, case in sorted(metrics["cases"].items(), key=lambda item: int(item[0])):
    step = int(step)
    print(f"### {step} µs")
    me = case["comparisons"]["me"]
    cs = case["comparisons"]["cs"]
    em = pd.read_csv(me["rolling_error_trace"])
    ec = pd.read_csv(cs["rolling_error_trace"])
    fig, ax = plt.subplots(2, 1, figsize=(11, 6), sharex=True)
    for frame, label in [(em, "DPsim ME"), (ec, "DPsim CS")]:
        ax[0].plot(frame.time_s, frame.pcc_voltage_rms_error_percent_100ms, label=label)
        ax[1].plot(
            frame.time_s, frame.inverter_current_rms_error_percent_100ms, label=label
        )
    ax[0].set_ylabel("PCC voltage RMS error [%]")
    ax[1].set_ylabel("Inverter current RMS error [%]")
    ax[1].set_xlabel("Time [s]")
    for a in ax:
        a.axvline(5, color="0.3", ls=":", label="25 MW load step")
        a.grid(True, alpha=0.3)
        a.legend()
    fig.suptitle(f"{step} µs: rolling 100 ms error vs native DPsim")
    plt.tight_layout()
    plt.show()

    native_path = case["runtimes"]["native"]["directory"] + "/native_trace.csv"
    native = pd.read_csv(native_path)
    fig, ax = plt.subplots(2, 1, figsize=(11, 6), sharex=True)
    for mode, label in (("me", "DPsim ME"), ("cs", "DPsim CS")):
        candidate = pd.read_csv(case["comparisons"][mode]["compact_trace"])
        if len(candidate) != len(native) or not np.allclose(
            candidate.time_s, native.time_s, rtol=0, atol=1e-12
        ):
            raise ValueError(
                f"{step} µs {label} compact trace does not align with native DPsim"
            )
        vcols = ["pcc_v_a_V", "pcc_v_b_V", "pcc_v_c_V"]
        icols = ["current_a_A", "current_b_A", "current_c_A"]
        vdiff = (candidate[vcols].to_numpy() - native[vcols].to_numpy()) ** 2
        idiff = (candidate[icols].to_numpy() - native[icols].to_numpy()) ** 2
        vref = native[vcols].to_numpy() ** 2
        iref = native[icols].to_numpy() ** 2
        verror = 100 * np.sqrt(
            np.cumsum(vdiff.sum(axis=1)) / np.cumsum(vref.sum(axis=1))
        )
        ierror = 100 * np.sqrt(
            np.cumsum(idiff.sum(axis=1)) / np.cumsum(iref.sum(axis=1))
        )
        ax[0].plot(native.time_s, verror, label=label)
        ax[1].plot(native.time_s, ierror, label=label)
    ax[0].set_ylabel("Cumulative PCC voltage RMS error [%]")
    ax[1].set_ylabel("Cumulative inverter current RMS error [%]")
    ax[1].set_xlabel("Time [s]")
    for a in ax:
        a.axvline(5, color="0.3", ls=":", label="25 MW load step")
        a.grid(True, alpha=0.3)
        a.legend()
    fig.suptitle(f"{step} µs: cumulative RMS error since t = 0")
    plt.tight_layout()
    plt.show()

    zm = pd.read_csv(me["load_step_zoom_trace"])
    zc = pd.read_csv(cs["load_step_zoom_trace"])
    fig, ax = plt.subplots(2, 1, figsize=(11, 6), sharex=True)
    ax[0].plot(zm.time_s, zm.native_v_a_V, "k--", label="Native DPsim")
    ax[0].plot(zm.time_s, zm.me_v_a_V, label="DPsim ME")
    ax[0].plot(zc.time_s, zc.cs_v_a_V, label="DPsim CS")
    ax[1].plot(zm.time_s, zm.native_i_a_A, "k--", label="Native DPsim")
    ax[1].plot(zm.time_s, zm.me_i_a_A, label="DPsim ME")
    ax[1].plot(zc.time_s, zc.cs_i_a_A, label="DPsim CS")
    ax[0].set_ylabel("Bus 3 phase-a voltage [V]")
    ax[1].set_ylabel("Inverter phase-a current [A]")
    ax[1].set_xlabel("Time [s]")
    for a in ax:
        a.grid(True, alpha=0.3)
        a.legend()
    fig.suptitle(f"{step} µs: 5 s load-step zoom")
    plt.tight_layout()
    plt.show()

    # Resolve the initial closed-loop transient that can be hidden in a 10 s view.
    native_path = case["runtimes"]["native"]["directory"] + "/native_trace.csv"
    native = pd.read_csv(native_path)
    fig, ax = plt.subplots(2, 1, figsize=(11, 6), sharex=True)
    ax[0].plot(native.time_s, native.pcc_v_a_V, "k--", label="Native DPsim")
    ax[1].plot(native.time_s, native.current_a_A, "k--", label="Native DPsim")
    for mode, label in (("me", "DPsim ME"), ("cs", "DPsim CS")):
        candidate = pd.read_csv(case["comparisons"][mode]["compact_trace"])
        ax[0].plot(candidate.time_s, candidate.pcc_v_a_V, label=label)
        ax[1].plot(candidate.time_s, candidate.current_a_A, label=label)
    cs_interface = pd.read_csv(cs["interface_trace"])
    ax[0].plot(
        cs_interface.time_s,
        cs_interface.fmi_input_v_a_V,
        ":",
        linewidth=1.0,
        label="CS FMU input (held voltage)",
    )
    ax[0].set_ylabel("Bus 3 phase-a voltage [V]")
    ax[1].set_ylabel("Inverter phase-a current [A]")
    ax[1].set_xlabel("Time [s]")
    for a in ax:
        a.set_xlim(0, 0.05)
        a.grid(True, alpha=0.3)
        a.legend()
    fig.suptitle(f"{step} µs: startup transient")
    plt.tight_layout()
    plt.show()
```

## Interpretation notes

Each timestep is a separate comparison and uses the native DPsim run at that same timestep as its baseline. The percentages describe implementation agreement at a matched discretization, not absolute physical accuracy. A coarse timestep changes the discrete-time network and inverter trajectories for all three methods. The 5000 µs case is a coarse-step stress test and should be interpreted separately.

The results show ME agreeing with native DPsim to numerical precision. CS starts with the same bus-3 voltage and current, but its electrical interface is a single-pass current source: it advances once per network step using the previously accepted terminal voltage, with no rollback or network Newton iteration. The generated CS input trace records this held-voltage exchange. The short-start diagnostics show that the full IEEE9 EMT network itself has a small alternating startup transient in native and ME; the explicit CS interface amplifies this mode and bus-3 voltage collapses within milliseconds. The FMU regulates active and reactive power by recalculating its current target from the terminal voltage, which creates a strong voltage-to-current feedback path. A one-pass held-voltage exchange does not solve that feedback implicitly, while ME does through network Newton iterations and the FMI current Jacobian. This explains why both CS voltage and current become physically implausible even though the FMU starts with correct states and port order. It is a CS coupling stability limitation for this inverter/grid combination, not evidence of a powerflow or FMI port-order mismatch. Treat CS data as a failed coupling case pending rollback-based iteration or a more stable interface method. Errors are relative to native DPsim, not absolute physical errors.
