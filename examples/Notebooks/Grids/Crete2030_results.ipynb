```python
import pandas as pd
df1 = pd.read_csv('/dpsim/examples/Notebooks/Grids/logs/2030_with_profile.csv', skipinitialspace=True)

df1.info()

print(df1.keys())
# 3️⃣ To list all column names
#print(df1.columns.tolist()) 

# Filter columns: keep only those that do NOT contain 'lne' or 'trf'
filtered_cols = df1.columns[~df1.columns.str.contains('lne|trf', case=False, regex=True)]

# Create a new DataFrame with only the filtered columns
df1_filtered = df1[filtered_cols]

# Optional: check which columns were kept
print(df1_filtered.columns.tolist())
```

```python
import matplotlib.pyplot as plt
import numpy as np

dataframes = [df1_filtered]



df1_filtered.plot()
plt.ylim(-1e9, 1e9)   # optional — limit y-axis
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')  # move legend outside
plt.show()
```
