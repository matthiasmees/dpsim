```python
import matplotlib.pyplot as plt
%matplotlib widget

gen_vars = {
    'i_gen1': ['GEN1.I_0', 'GEN1.I_1', 'GEN1.I_2'],
    'i_gen2': ['GEN2.I_0', 'GEN2.I_1', 'GEN2.I_2'],
    'i_gen3': ['GEN3.I_0', 'GEN3.I_1', 'GEN3.I_2'],
}


dataframes = [df1, df2, df3, df4, df5]
titles = ['Sim']

fig, axes = plt.subplots(1, 3, figsize=(12, 3))  # 1 row, 3 columns

df = dataframes[0]  # Since you only have one dataframe

for ax, (gen_name, var_list) in zip(axes, gen_vars.items()):
    for var in var_list:
        ax.plot(df['time'].values, df[var].values, linestyle="--", label=var.split('_')[-1])
    ax.set_title(gen_name)
    ax.set_xlabel('Time')
    ax.set_ylabel('Current')
    ax.grid(True)
    ax.legend(title="Phase")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

dataframes = [df1, df2, df3, df4, df5]
phases = ['v6_0', 'v6_1', 'v6_2']
num_runs = len(dataframes)


fig, axes = plt.subplots(nrows=num_runs, ncols=len(phases), 
                         figsize=(15, 3*num_runs), sharex=True)


for i, df in enumerate(dataframes):
    for j, phase in enumerate(phases):
        ax = axes[i, j]
        ax.plot(df['time'], df[phase])   # <-- use time on x-axis
        ax.set_title(f"EMT DPsim Sim {i+1} - {phase}")
        ax.set_xlabel("Time [s]")
        ax.set_ylabel("Voltage [V]")
        ax.grid(True)

plt.tight_layout()
plt.show()

### RUN 1 - 33% Load, RUN 2 - 66% Load, RUN 3 - 100% Load, RUN 4 - 133% Load, RUN 5 - 166% Load ###
```

```python
import matplotlib.pyplot as plt
%matplotlib widget

gen_vars = {
    'GEN1.I_0', 'GEN1.I_1', 'GEN1.I_2',
    'GEN2.I_0', 'GEN2.I_1', 'GEN2.I_2',
    'GEN3.I_0', 'GEN3.I_1', 'GEN3.I_2',
    'GEN1.V_0', 'GEN1.V_1', 'GEN1.V_2',
    'GEN2.V_0', 'GEN2.V_1', 'GEN2.V_2',
    'GEN3.V_0', 'GEN3.V_1', 'GEN3.V_2'
}

dataframes = [df1, df2, df3, df4, df5]
titles = ['Sim']
generators = [1, 2, 3]

fig, axes = plt.subplots(nrows=5, ncols=3, figsize=(15, 15), sharex=True)


for i, df in enumerate(dataframes):
    for j, gen in enumerate(generators):
        
        P = df[f"GEN{gen}.V_0"]*df[f"GEN{gen}.I_0"] + df[f"GEN{gen}.V_1"]*df[f"GEN{gen}.I_1"] + df[f"GEN{gen}.V_2"]*df[f"GEN{gen}.I_2"]
        
        ax = axes[i, j]
        ax.plot(df["time"], P / 1e8)
        ax.set_title(f"Sim {i+1} - GEN{gen} Power")
        ax.set_ylabel("Power [pu]")
        ax.grid(True)
        
        # Only label x-axis on bottom row
        if i == 4:
            ax.set_xlabel("Time [s]")

plt.tight_layout()
plt.show()
```
