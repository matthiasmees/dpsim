```python
import matplotlib.pyplot as plt

dataframes = [df1, df2, df3, df4, df5]
dataframesDPsim = [df1dpsim, df2dpsim, df3dpsim, df4dpsim, df5dpsim]
phases = ['V6A', 'V6B', 'V6C']
phasesDPsim = ['v6_0', 'v6_1', 'v6_2']

num_runs = len(dataframes)

fig, axes = plt.subplots(nrows=num_runs, ncols=len(phases), 
                         figsize=(15, 3*num_runs), sharex=True)



for i, df in enumerate(dataframes):
    for j, phase in enumerate(phases):
        ax = axes[i, j]
        ax.plot(df['Time'], df[phase] * 1000)   # <-- use time on x-axis
        ax.set_title(f"RSCAD Sim {i+1} - {phase}")
        ax.set_xlabel("Time [s]")
        ax.set_ylabel("Voltage [V]")
        ax.grid(True)

plt.tight_layout()
plt.show()

### RUN 1 - 33% Load, RUN 2 - 66% Load, RUN 3 - 100% Load, RUN 4 - 133% Load, RUN 5 - 166% Load ###
```
