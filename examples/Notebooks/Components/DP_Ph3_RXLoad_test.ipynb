```python
print(df1.keys())
```

```python

gen_vars_re = {
    'i12re': ['i12_0.re','i12_1.re','i12_2.re'],
    'v1re': ['v1_0.re','v1_1.re','v1_2.re'],
    'v2re': ['v2_0.re','v2_1.re','v2_2.re']
}



dataframes = [df1]
titles = ['Sim']

fig, axes = plt.subplots(1, 3, figsize=(12, 3))  # 1 row, 3 columns

df = dataframes[0]  # Since you only have one dataframe

for ax, (gen_name, var_list) in zip(axes, gen_vars_re.items()):
    for var in var_list:
        ax.plot(df['time'].values, df[var].values, linestyle="--", label=var.split('_')[-1])
    ax.set_title(gen_name)
    ax.set_xlabel('Time')
    ax.set_ylabel('Current/Voltage')
    ax.grid(True)
    ax.legend(title="Phase")

plt.tight_layout()
plt.show()
```
